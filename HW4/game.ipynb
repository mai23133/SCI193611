# HW4 — Hexapawn vs Alpha-Beta AI

This notebook uses `game.py`, which implements standard 3×3 Hexapawn. White starts at rank 1 and moves upward; Black starts at rank 3 and moves downward. A player wins by reaching the opposite back rank or leaving the opponent with no legal move.

```python
from game import Hexapawn, alpha_beta_player, format_move, launch_gui

game = Hexapawn()
state = game.initial
game.display(state)
```

## Alpha-Beta move
`alpha_beta_player(game, state)` has the same callback form as the AIMA TicTacToe player. It searches the complete, small Hexapawn game tree and returns the optimal legal move.

```python
move = alpha_beta_player(game, state)
print('Alpha-Beta chooses:', format_move(move))
state = game.result(state, move)
game.display(state)
```
